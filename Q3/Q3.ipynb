# Q3 – Engineer a Vehicle Registration Number Validator

**Aim:** To develop a Python program that normalises vehicle registration numbers, validates their structure using string operations, and formats valid numbers for receipts.

**Valid registration format:**
- 2 letters – State
- 2 digits – District
- 1 or 2 letters – Series
- 4 digits – Registration number

Regular expressions are not used in this program.

### Prediction Before Execution

| Input | Predicted result |
|---|---|
| tn 09 ab 1234 | Valid |
| KA-01-M-0007 | Valid |
| Tn09Ab1234 | Valid |
| TN9AB1234 | Invalid |
| TN 22 1234 AB | Invalid |
| MH-12-AB-123 | Invalid |

I expect the program to remove spaces and hyphens, convert letters into uppercase, and reject numbers that do not follow the required structure.

````python
# Q3 - Function Definitions

def normalise(text):
    text = text.strip()
    text = text.replace(" ", "")
    text = text.replace("-", "")
    text = text.upper()
    return text

def is_valid(reg):
    if len(reg) != 9 and len(reg) != 10:
        return False

    state = reg[0:2]
    district = reg[2:4]
    series = reg[4:-4]
    number = reg[-4:]

    if not state.isalpha():
        return False

    if not district.isdigit():
        return False

    if not series.isalpha():
        return False

    if not number.isdigit():
        return False

    return True

def format_receipt(reg):
    if is_valid(reg):
        return reg[0:2] + " " + reg[2:4] + " " + reg[4:-4] + " " + reg[-4:]
    return "INVALID REGISTRATION"
````

````python
# Test the functions with the predicted inputs
test_cases = [
    "tn 09 ab 1234",
    "KA-01-M-0007",
    "Tn09Ab1234",
    "TN9AB1234",
    "TN 22 1234 AB",
    "MH-12-AB-123"
]

for raw in test_cases:
    reg = normalise(raw)
    valid = is_valid(reg)
    result = format_receipt(reg) if valid else "INVALID"
    print(f"Raw: {raw:<16} -> Normalised: {reg:<12} -> {'Valid' if valid else 'Invalid':<8} -> Receipt: {result}")
````

````python
# Q3 - Six Registration Number Tests

test_inputs = [
    "tn 09 ab 1234",
    "KA-01-M-0007",
    "Tn09Ab1234",
    "TN9AB1234",
    "TN 22 1234 AB",
    "MH-12-AB-123"
]

print("REGISTRATION VALIDATION RESULTS")

for raw in test_inputs:
    reg = normalise(raw)
    if is_valid(reg):
        print(raw, "->", format_receipt(reg), "VALID")
    else:
        print(raw, "->", reg, "INVALID")
````

### Q3 – Analysis and Written Answers

**1. Why are strings immutable in Python?**

Strings are immutable because their characters cannot be changed after the string is created. Any modification creates a new string instead of changing the original string.

**2. Why does text.upper() not change text?**

The upper() method returns a new string containing uppercase characters. It does not modify the original string. To save the result, we must assign it back to the variable.

Example:

```python
text = "hello"
text.upper()
print(text)  # hello
```

**3. What does s[-4:] return?**

It returns the last four characters of the string. These represent the four-digit registration number.

**4. What does s[4:-4] return?**

It returns the series letters between the district digits and the final four digits.

For a 9-character registration:

- s[-4:] returns the last four digits.
- s[4:-4] returns one series letter.

For a 10-character registration:

- s[-4:] returns the last four digits.
- s[4:-4] returns two series letters.

**5. Alternative concept – Regular expressions**

A regular expression can validate the registration using:

```python
import re

def is_valid_regex(reg):
    return re.fullmatch(
            r"[A-Z]{2}[0-9]{2}[A-Z]{1,2}[0-9]{4}",
                    reg
                        ) is not None
                        ```

                        **6. Complexity analysis**

                        - String slicing and validation: O(L) time and O(L) auxiliary space for normalisation.
                        - Regular expression validation: O(L) time for this fixed pattern and O(1) auxiliary space, excluding implementation-specific internal allocations.

                        Here, L represents the length of the registration number.

                        String slicing is easy to understand for this fixed format. Regular expressions are shorter and can be easier to maintain when validation patterns become more complicated.
