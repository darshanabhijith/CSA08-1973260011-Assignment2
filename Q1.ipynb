Q1 - Decisions with Many Boundaries
Electricity Subsidy Eligibility

### Prediction Before Execution – Q1

I predict the following results for eight households based on their electricity consumption.

| Household | Units | Ration Card | Member Above 80 | Expected Result |
|---|---:|---|---|---|
| H1 | 50 | Yes | No | Full subsidy |
| H2 | 100 | No | No | Full subsidy |
| H3 | 101 | Yes | No | Partial subsidy |
| H4 | 150 | No | No | No subsidy |
| H5 | 200 | Yes | No | Partial subsidy |
| H6 | 201 | Yes | No | No subsidy |
| H7 | 250 | No | Yes | Subsidy under age exception |
| H8 | 300 | Yes | Yes | Subsidy under age exception |

```python
# Q1 - Electricity Subsidy Eligibility

households = [
    (50, True, False),
    (100, False, False),
    (101, True, False),
    (150, False, False),
    (200, True, False),
    (201, True, False),
    (250, False, True),
    (300, True, True)
]

# Method 1: Separate if statements

print("USING SEPARATE IF STATEMENTS")

for units, ration, senior in households:
    print("\nUnits:", units)

    # Note: Households <= 100 get Full subsidy, but if 101 <= units <= 200 it depends on the ration card.
    if units <= 200:
        print("Full subsidy")

    if 101 <= units <= 200 and ration:
        print("Partial subsidy")

    if units > 200 and senior:
        print("Special subsidy")

    if units > 200 and not senior:
        print("No subsidy")

    if 101 <= units <= 200 and not ration:
        print("No subsidy")


# Method 2: if-elif-else

print("\nUSING IF-ELIF-ELSE")

for units, ration, senior in households:
    if units <= 100:
        result = "Full subsidy"
    elif 101 <= units <= 200:
        if ration:
            result = "Partial subsidy"
        else:
            result = "No subsidy"
    else:
        if senior:
            result = "Special subsidy"
        else:
            result = "No subsidy"

    print("Units:", units, "->", result)
```

### Q1 – Analysis and Design Decision

**1. Why can separate if statements give more than one result?**

Separate if statements check every condition independently. If multiple conditions are true, more than one result can be printed. This can create incorrect subsidy decisions.

**2. Why does if-elif-else give exactly one result?**

The if-elif-else structure checks conditions in order. Once one condition becomes true, the remaining conditions are skipped. Therefore, only one result is selected.

**3. Nested if or combined Boolean conditions?**

I prefer nested if statements because they make the program easier to understand. First, the electricity units are checked, and then additional conditions such as ration card ownership or senior citizen status are checked.

**4. Alternative concept – Match-case**

Python's match-case statement can also be used to implement the same decision:

- Case 1: Units are 100 or less – Full subsidy.
- Case 2: Units are between 101 and 200 – Check ration card.
- Case 3: Units are above 200 – Check whether a member is above 80.
- Default case: No subsidy.

**5. Complexity analysis**

- If-elif-else: O(1) time and O(1) space for a fixed number of conditions.
- Match-case: O(1) time and O(1) space for a fixed number of cases.
- Lookup list: O(k) time in the worst case and O(k) space, where k is the number of slabs.

Both if-elif-else and match-case are suitable for this problem. Match-case requires Python 3.10 or later.

```python
# Q1 - Alternative Method Using match-case

def subsidy_match(units, ration, senior):
    match units:
        case n if n <= 100:
            return "Full subsidy"
        case n if 101 <= n <= 200:
            if ration:
                return "Partial subsidy"
            else:
                return "No subsidy"
        case n if n > 200:
            if senior:
                return "Special subsidy"
            else:
                return "No subsidy"

print("USING MATCH-CASE")
for units, ration, senior in households:
    result = subsidy_match(units, ration, senior)
    print("Units:", units, "->", result)
```

```python
# Q1 - Alternative Method Using Lookup List

def subsidy_lookup(units, ration, senior):
    slabs = [
        (100, "FULL"),
        (200, "MIDDLE"),
        (float("inf"), "ABOVE")
    ]

    for limit, category in slabs:
        if units <= limit:
            if category == "FULL":
                return "Full subsidy"
            elif category == "MIDDLE":
                if ration:
                    return "Partial subsidy"
                else:
                    return "No subsidy"
            else:
                if senior:
                    return "Special subsidy"
                else:
                    return "No subsidy"

print("USING LOOKUP LIST")
for units, ration, senior in households:
    result = subsidy_lookup(units, ration, senior)
    print("Units:", units, "->", result)
```
