Aim: To understand Python functions, parameters, default arguments and variable scope using a college canteen order system.

Prediction: I expect the function with order=[] to incorrectly share the same list between customers. Using order=None should create a separate list for each customer.

```python
# Q5 - Default Parameter Demonstration

def add_item_buggy(item, order=[]):
    order.append(item)
    return order

def add_item(item, order=None):
    if order is None:
        order = []
    order.append(item)
    return order

print("BUGGY FUNCTION:")
print(add_item_buggy("Tea"), add_item_buggy("Vada"))

print("\nCORRECT FUNCTION:")
print(add_item("Tea"), add_item("Vada"))

print("\nDEFAULT VALUE:")
print(add_item_buggy.__defaults__)
```

```python
# Q5 - Three Customer Orders

customer1 = add_item("Tea")
customer1 = add_item("Vada", customer1)

customer2 = add_item("Coffee")
customer2 = add_item("Sandwich", customer2)

customer3 = add_item("Juice")
customer3 = add_item("Biscuit", customer3)

print("Customer 1:", customer1)
print("Customer 2:", customer2)
print("Customer 3:", customer3)
```

```python
# Positional, Keyword and Default Parameters

def order_details(item, quantity=1):
    print(item, quantity)

order_details("Tea", 2)
order_details(item="Coffee", quantity=3)
order_details("Juice")


# Returning Multiple Values

def calculate_bill(amount):
    gst = amount * 0.05
    total = amount + gst
    return total, gst

total, gst = calculate_bill(100)

print("GST:", gst)
print("Total:", total)


# Local and Global Scope

canteen_name = "College Canteen"

def show_canteen():
    message = "Welcome"
    print(message)
    print(canteen_name)

show_canteen()


# Nonlocal Scope

def outer_function():
    count = 0

    def inner_function():
        nonlocal count
        count += 1

    inner_function()
    print("Nonlocal count:", count)

outer_function()


# Global Keyword

visitors = 0

def add_visitor():
    global visitors
    visitors += 1

add_visitor()
print("Visitors:", visitors)
```

1. Why does the default list cause a problem?

A default list is created only once when the function is defined. Every function call shares the same list, so items from previous customers appear in new orders.

2. How does order=None solve the problem?

When the argument is None, the function creates a new empty list. This ensures that every customer gets an independent order.

3. When are positional, keyword and default parameters used?
- Positional parameters are passed according to their position.
- Keyword parameters are passed using parameter names.
- Default parameters provide a value when the caller does not supply one.
4. How can a function return multiple values?

A Python function can return multiple values separated by commas. They can be stored in separate variables.

5. Explain local, global and nonlocal variables.
- Local variables are created inside a function and are normally available only there.
- Global variables are defined outside functions and can be accessed throughout the module.
- Nonlocal variables refer to variables in an enclosing function.
6. Why is the global keyword needed?

The global keyword allows a function to modify a global variable. Excessive use of global variables is avoided because it makes programs harder to maintain and debug.

7. What is the alternative to using order=None?

The caller can always pass a separate list explicitly. Another option is using a closure that maintains a separate order list. Both approaches can keep customer orders independent.
6. Complexity analysis
- Adding one item to a Python list takes O(1) amortised time.
- Each customer's order requires O(m) space for m items.
- Using order=None creates a separate list for each new customer.
- Passing a list explicitly also takes O(1) amortised time per added item.
