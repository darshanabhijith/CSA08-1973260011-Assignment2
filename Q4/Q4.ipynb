Aim: To compare three methods of building a long SMS statement: repeated +=, list with join(), and f-strings.

Prediction: I expect join() to be faster for large statements because it combines the parts efficiently in one operation. Repeated += may take longer because strings are immutable, so new strings may need to be created during concatenation.

```python
import time

items = [f"Item{i}:Rs{i % 50}" for i in range(200_000)]

def build_plus(parts):
    text = ""
    for p in parts:
        text += p + ";"
    return text

def build_join(parts):
    return "".join([p + ";" for p in parts])

def build_fstring(parts):
    lines = [f"{p};" for p in parts]
    return "".join(lines)

start = time.perf_counter()
a = build_plus(items)
t_plus = time.perf_counter() - start

start = time.perf_counter()
b = build_join(items)
t_join = time.perf_counter() - start

start = time.perf_counter()
c = build_fstring(items)
t_fstring = time.perf_counter() - start

print("Same text:", a == b == c)
print("Length:", len(b))
print(f"+= time: {t_plus:.4f} seconds")
print(f"join time: {t_join:.4f} seconds")
print(f"f-string time: {t_fstring:.4f} seconds")
```

1. Why are strings immutable in Python?

Strings are immutable because their characters cannot be changed after creation. When we concatenate strings, Python creates a new string instead of modifying the original one.

2. Why can repeated += become slow?

Repeated += can create new strings during concatenation. As the statement grows, more characters may need to be copied, which can increase processing time.

3. Why is join() recommended?

join() combines multiple string parts efficiently into one string. It is generally preferred when building large text because it avoids repeated concatenation.

4. Why can the += result sometimes be faster than expected?

CPython sometimes optimises repeated += operations. Therefore, actual timings can vary depending on the Python implementation and computer.

5. What is the alternative using io.StringIO or a file?

io.StringIO allows text to be written into an in-memory text stream using write(). A file can also be used to write each statement line directly, avoiding the need to keep the entire statement in one growing string.

6. Complexity analysis
- Repeated +=: can require up to O(n²) character copying in the general case.
- join(): O(n) time in the total output size, with memory for the parts and final string.
- F-string with join(): O(n) time in the total output size.
- StringIO or file writing: generally O(n) total writing time; a file avoids keeping the complete output string in memory.
Here, n represents the total amount of text being built.
