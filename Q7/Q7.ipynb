# Q7 – Day-4 Design Challenge: Diagnose, Redesign and Validate

## Selected Program
Q6 – Counting Ways to Climb Temple Steps

## A. Diagnose the Bottleneck

The original recursive solution repeats calculations for the same smaller step counts. For n = 25, plain recursion made 242,785 function calls, while memoisation reduced this to 26 calls.

However, memoisation still uses recursive function calls. When tested with n = 1,000, it raised a RecursionError because Python has a recursion-depth limit.

## B. Design Objective

The objective is to calculate the number of ways for up to 1,000 steps without a recursion-depth error, while reducing extra memory usage to O(1).

## C. Revised Design

I will use an iterative approach. The program stores only the previous two results and calculates the next result using a loop.

Pseudocode:

1. If n is 0 or 1, return 1.
2. Set previous = 1 and current = 1.
3. Repeat from step 2 to n:
   - Calculate next = previous + current.
      - Move current into previous.
         - Store next in current.
         4. Return current.

         The iterative design avoids recursive calls and does not need a cache. Its trade-off is that it does not retain previously calculated answers for reuse in other calculations.

         ## D. Prediction Before Implementation

         I predict that the iterative solution will handle 1,000 steps successfully, avoid recursion errors and use O(1) extra space. It should also perform fewer operations than plain recursion for larger inputs.

```python
from functools import lru_cache
import time

memo_calls = 0

@lru_cache(maxsize=None)
def ways_memo(n):
    global memo_calls
    memo_calls += 1
    if n <= 1:
        return 1
    return ways_memo(n - 1) + ways_memo(n - 2)

# Test the original solution
print("ORIGINAL MEMOISED SOLUTION")

for n in [10, 20, 25, 30]:
    ways_memo.cache_clear()
    memo_calls = 0
    start = time.perf_counter()
    result = ways_memo(n)
    elapsed = time.perf_counter() - start
    print("Steps:", n)
    print("Result:", result)
    print("Function calls:", memo_calls)
    print("Time:", round(elapsed, 6), "seconds")
    print()
```

```python
from functools import lru_cache
import time

# Q7 - Redesigned Iterative Solution

memo_calls = 0

@lru_cache(maxsize=None)
def ways_memo(n):
    global memo_calls
    memo_calls += 1
    if n <= 1:
        return 1
    return ways_memo(n - 1) + ways_memo(n - 2)

def ways_iterative(n):
    if n <= 1:
        return 1, 0

    previous = 1
    current = 1
    iterations = 0

    for i in range(2, n + 1):
        previous, current = current, previous + current
        iterations += 1

    return current, iterations

# Compare both solutions using the same inputs
print("ORIGINAL vs REDESIGNED")

for n in [10, 20, 25, 30, 1000]:
    ways_memo.cache_clear()
    memo_calls = 0
    print("\nSteps:", n)

    try:
        start = time.perf_counter()
        original_result = ways_memo(n)
        original_time = time.perf_counter() - start
        print("Memoised result:", original_result)
        print("Memoised calls:", memo_calls)
        print("Memoised time:", round(original_time, 6), "seconds")
    except RecursionError:
        print("Memoised solution: RecursionError")

    start = time.perf_counter()
    new_result, iterations = ways_iterative(n)
    new_time = time.perf_counter() - start
    print("Iterative result:", new_result)
    print("Iterative iterations:", iterations)
    print("Iterative time:", round(new_time, 6), "seconds")
    print("Number of digits:", len(str(new_result)))
```

## E. Implement and Validate

The original memoised solution and the redesigned iterative solution were tested using the same inputs.

For 10, 20, 25 and 30 steps, both solutions returned identical results. At 25 steps, memoisation required 26 function calls, while the iterative solution used 24 loop iterations.

At 1,000 steps, the memoised solution raised a RecursionError. The iterative solution completed successfully using 999 iterations and produced a 209-digit result.

The measurements support the prediction that the iterative approach can handle larger inputs without reaching Python's recursion-depth limit.

## F. Defend the Final Design

The redesign achieved its objective. It calculates the number of ways for 1,000 steps without recursion and uses only O(1) extra working space.

The iterative solution has O(n) time complexity and O(1) auxiliary space. Memoisation also has O(n) time complexity, but requires O(n) cache space and recursive stack space.

The iterative approach is preferable for large inputs because it avoids recursion-depth errors and uses less additional memory. Memoisation remains useful when previously calculated results need to be reused.

## Final Conclusion

The iterative redesign is more suitable for calculating climbing combinations for up to 1,000 steps. The experiment confirms that selecting an appropriate algorithm can improve reliability and memory efficiency as input size increases.
