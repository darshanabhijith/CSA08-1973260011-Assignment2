Aim: To calculate the number of ways to climb temple steps by taking either one or two steps at a time, using plain recursion, memoisation and iteration.

Prediction: Plain recursion will make many repeated function calls. Memoisation will reduce repeated calculations by storing previous results. An iterative loop should also take O(n) time and use O(1) extra space.

For five steps, the recursive relationship is:
\[
ways(n)=ways(n-1)+ways(n-2)
\]

Base cases:
\[
ways(0)=1,\quad ways(1)=1
\]

Call tree:
ways(5)
├── ways(4)
│   ├── ways(3)
│   │   ├── ways(2)
│   │   └── ways(1)
│   └── ways(2)
└── ways(3)
    ├── ways(2)
        └── ways(1)

```python
from functools import lru_cache
import sys

calls = 0

def ways_naive(n):
    global calls
    calls += 1
    if n <= 1:
        return 1
    return ways_naive(n - 1) + ways_naive(n - 2)

@lru_cache(maxsize=None)
def ways_memo(n):
    global calls
    calls += 1
    if n <= 1:
        return 1
    return ways_memo(n - 1) + ways_memo(n - 2)

# Correctness check: n = 1 to 6
print("CORRECTNESS CHECK")
for n in range(1, 7):
    print("n =", n, "ways =", ways_memo(n))

# Required comparison for n = 25
calls = 0
print("\nNaive:", ways_naive(25), "calls =", calls)

calls = 0
ways_memo.cache_clear()
print("Memo :", ways_memo(25), "calls =", calls)
```

```python
print("FUNCTION CALL COMPARISON")

for n in [10, 20, 25, 30]:
    calls = 0
    naive_result = ways_naive(n)
    naive_calls = calls

    calls = 0
    ways_memo.cache_clear()
    memo_result = ways_memo(n)
    memo_calls = calls

    print("\nn =", n)
    print("Naive result:", naive_result)
    print("Naive calls:", naive_calls)
    print("Memo result:", memo_result)
    print("Memo calls:", memo_calls)
```

```python
print("RECURSION LIMIT:", sys.getrecursionlimit())

ways_memo.cache_clear()

try:
    print("Memoised result for 1000:", ways_memo(1000))
except RecursionError:
    print("RecursionError: recursion depth limit reached")

def ways_loop(n):
    if n <= 1:
        return 1

    previous = 1
    current = 1

    for i in range(2, n + 1):
        previous, current = current, previous + current

    return current

print("\nITERATIVE RESULT")
print("Ways for 1000 steps:", ways_loop(1000))
print("Number of digits:", len(str(ways_loop(1000))))
```

1. What is the base case?

The base cases are ways(0) = 1 and ways(1) = 1. There is one way to climb zero steps and one way to climb one step.

2. What is the recursive case?

The recursive case is ways(n) = ways(n-1) + ways(n-2). A person can reach the final step by taking either one step or two steps.

3. Why is plain recursion inefficient?

Plain recursion calculates the same smaller problems many times. This causes unnecessary function calls and increases execution time.

4. How does memoisation help?

Memoisation stores previously calculated results in a cache. When the same value is needed again, Python reuses the stored answer instead of calculating it again.

5. Why does memoisation still have a recursion limit?

Memoisation reduces repeated calculations, but it still uses recursive function calls. Python has a maximum recursion depth. A large input such as 1,000 can cause a RecursionError.

6. What is the iterative alternative?

An iterative solution uses a loop and stores only the previous two results. It avoids recursive function calls and can calculate the answer for 1,000 steps.

7. Complexity analysis
Method	Time complexity	Extra space
Plain recursion	O(φⁿ)	O(n) stack
Memoisation	O(n)	O(n) cache and stack
Iterative loop	O(n)	O(1)

Here, φ represents the golden ratio, approximately 1.618.

8. Deployment decision

For a staircase of up to 1,000 steps, I would choose the iterative solution. It avoids recursion-depth errors, requires constant extra space and calculates the answer efficiently. Memoisation is useful for demonstrating recursion and avoiding repeated calculations, but the iterative approach is more suitable for this larger input.
