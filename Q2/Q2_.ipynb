```python
# Q2 - Loops that Count Attempts
# OTP Verification at a Bank Kiosk

CORRECT_OTP = "2468"


# METHOD 1: WHILE LOOP

def otp_while(entries):
    print("\n--- WHILE LOOP ---")
    attempts = 0
    index = 0

    while attempts < 3 and index < len(entries):
        otp = entries[index]
        index += 1

        if otp == "":
            print("Blank entry - not counted")
            continue

        attempts += 1
        print("Attempt", attempts, ":", otp)

        if otp == CORRECT_OTP:
            print("Login successful!")
            break
        elif otp == "0000":
            pass
        else:
            print("Incorrect OTP")
    else:
        if attempts == 3:
            print("Account locked for 30 minutes")
        else:
            print("No more entries available")


# METHOD 2: FOR LOOP WITH RANGE(3)

def otp_for(entries):
    print("\n--- FOR LOOP ---")
    attempts = 0
    index = 0

    for i in range(3):
        while index < len(entries):
            otp = entries[index]
            index += 1

            if otp == "":
                print("Blank entry - not counted")
                continue
            break
        else:
            print("No more entries available")
            break

        attempts += 1
        print("Attempt", attempts, ":", otp)

        if otp == CORRECT_OTP:
            print("Login successful!")
            break
        elif otp == "0000":
            pass
        else:
            print("Incorrect OTP")
    else:
        print("Account locked for 30 minutes")


# TEST 1: CORRECT OTP

success_entries = ["1111", "", "2468"]

otp_while(success_entries)
otp_for(success_entries)


# TEST 2: THREE WRONG ATTEMPTS

wrong_entries = ["1111", "", "2222", "3333"]

otp_while(wrong_entries)
otp_for(wrong_entries)


# CHANGED CONDITION:
# TWO EXTRA ATTEMPTS AFTER CORRECT SECURITY ANSWER

def otp_security(entries, security_correct):
    print("\n--- SECURITY QUESTION ---")
    if security_correct:
        maximum = 5
        print("Security answer correct. Two extra attempts granted.")
    else:
        maximum = 3
        print("Security answer incorrect. Three attempts allowed.")

    attempts = 0
    for otp in entries:
        if attempts == maximum:
            break

        if otp == "":
            print("Blank entry - not counted")
            continue

        attempts += 1
        print("Attempt", attempts, ":", otp)

        if otp == CORRECT_OTP:
            print("Login successful!")
            break
        else:
            print("Incorrect OTP")
    else:
        if attempts == maximum:
            print("Account locked for 30 minutes")
        else:
            print("No more entries available")


otp_security(["1111", "2222", "3333", "4444", "2468"], True)


# ALTERNATIVE METHOD: RECURSION

def otp_recursive(entries, index=0, attempts=0):
    if attempts == 3:
        print("Recursive: Account locked")
        return

    if index >= len(entries):
        print("Recursive: No more entries")
        return

    otp = entries[index]

    if otp == "":
        print("Recursive: Blank entry not counted")
        return otp_recursive(entries, index + 1, attempts)

    print("Recursive attempt:", attempts + 1)

    if otp == CORRECT_OTP:
        print("Recursive: Login successful!")
        return

    print("Recursive: Incorrect OTP")
    return otp_recursive(entries, index + 1, attempts + 1)


print("\n--- RECURSIVE METHOD ---")
otp_recursive(["1111", "", "2222", "3333"])
```
